```python
import ROOT
```

## Beautifying Plots

```python
DATA_SET = "v8_sample_filtered"

f_theta = ROOT.TFile(f"plots/{DATA_SET}/compare_theta.root")
f_theta.ls()

f_phi = ROOT.TFile(f"plots/{DATA_SET}/compare_phi.root")
f_phi.ls()
```

```python
STYLE_FILE = "src/style.C"
ROOT.gROOT.ProcessLine(f'.L {STYLE_FILE}')
ROOT.SetReleaseNoteStyle()
ROOT.DefineColors()
```

```python
h_r = f_theta.Get("reco_hist")
h_t = f_theta.Get("truth_hist")

print(f"reconstructed: {h_r.GetEntries()}, Truth: {h_t.GetEntries()}, reco/truth: {h_r.GetEntries()/h_t.GetEntries() if h_t.GetEntries() != 0 else 0}")
```

```python
# Match plot_prepped_data styling for compare_theta.
ROOT.SetReleaseNoteStyle()
ROOT.DefineColors()

for hist, color, marker in ((h_r, ROOT.BLUE, 20), (h_t, ROOT.MAGENTA, 24)):
    hist.SetLineColor(color)
    hist.SetFillColorAlpha(color, 0.7)
    hist.SetMarkerColor(color)
    hist.SetMarkerStyle(marker)
    hist.SetMarkerSize(0.7)
    hist.SetLineWidth(2)

for hist in (h_r, h_t):
    integral = hist.Integral("width")
    if integral > 0:
        hist.Scale(1.0 / integral)

h_r.SetTitle("Reconstructed vs Generated Polar Angle")
h_r.GetXaxis().SetTitle("#theta [rad]")
h_r.GetYaxis().SetTitle("Normalized entries")

theta_fd_min = 0.087
theta_fd_max = 0.611
c_theta = ROOT.TCanvas("c_compare_theta", "pflow vs truth theta", 1200, 900)
h_r.Draw("hist")
h_t.Draw("hist same")

y_max = 1.05 * max(h_r.GetMaximum(), h_t.GetMaximum())
fd_line_color = ROOT.TColor.GetColor(150, 150, 150)
fd_lines = []
for theta in (theta_fd_min, theta_fd_max):
    line = ROOT.TLine(theta, 0.0, theta, y_max)
    line.SetLineColorAlpha(fd_line_color, 0.55)
    line.SetLineStyle(2)
    line.SetLineWidth(2)
    line.Draw()
    fd_lines.append(line)

fd_label = ROOT.TLatex()
fd_label.SetTextAlign(22)
fd_label.SetTextColorAlpha(fd_line_color, 0.75)
fd_label.SetTextSize(0.035)
fd_label.DrawLatex(0.5 * (theta_fd_min + theta_fd_max), 0.98 * y_max, "Forward Detector (FD)")

legend = ROOT.TLegend(0.6, 0.7, 0.90, 0.90)
legend.SetBorderSize(0)
legend.AddEntry(h_r, "CLAS12 Recon", "f")
legend.AddEntry(h_t, "RGA Clasdis Gen", "f")
legend.Draw()

c_theta.Draw()
c_theta.SaveAs("compare_theta_notebook.pdf")
```

```python
h_phi_r = f_phi.Get("reco_hist")
h_phi_t = f_phi.Get("truth_hist")

print(f"reconstructed: {h_phi_r.GetEntries()}, Truth: {h_phi_t.GetEntries()}, reco/truth: {h_phi_r.GetEntries()/h_phi_t.GetEntries() if h_phi_t.GetEntries() != 0 else 0}")
```

```python
# Match plot_prepped_data styling for compare_phi.
ROOT.SetReleaseNoteStyle()
ROOT.DefineColors()

for hist, color, marker in ((h_phi_r, ROOT.BLUE, 20), (h_phi_t, ROOT.MAGENTA, 24)):
    hist.SetLineColor(color)
    hist.SetFillColorAlpha(color, 0.7)
    hist.SetMarkerColor(color)
    hist.SetMarkerStyle(marker)
    hist.SetMarkerSize(0.7)
    hist.SetLineWidth(2)

for hist in (h_phi_r, h_phi_t):
    integral = hist.Integral("width")
    if integral > 0:
        hist.Scale(1.0 / integral)

h_phi_r.SetTitle("Reconstructed vs Generated Azimuthal Angle")
h_phi_r.GetXaxis().SetTitle("#font[1]{#phi} [rad]")
h_phi_r.GetYaxis().SetTitle("Normalized entries")
h_phi_r.GetYaxis().SetRangeUser(0,0.3)

c_phi = ROOT.TCanvas("c_compare_phi", "pflow vs truth phi", 1200, 900)
h_phi_r.Draw("hist")
h_phi_t.Draw("hist same")

legend_phi = ROOT.TLegend(0.65, 0.75, 0.90, 0.90)
legend_phi.SetBorderSize(0)
legend_phi.AddEntry(h_phi_r, "CLAS12 Recon", "f")
legend_phi.AddEntry(h_phi_t, "RGA Clasdis Gen", "f")
legend_phi.Draw()

c_phi.Draw()
c_phi.SaveAs("compare_phi_notebook.pdf")
```

```python
# Compare theta and phi distributions side by side.
ROOT.SetReleaseNoteStyle()
ROOT.DefineColors()

h_theta_r = f_theta.Get("reco_hist")
h_theta_t = f_theta.Get("truth_hist")
h_phi_r = f_phi.Get("reco_hist")
h_phi_t = f_phi.Get("truth_hist")

for hist, color, marker in (
    (h_theta_r, ROOT.BLUE, 20),
    (h_theta_t, ROOT.MAGENTA, 24),
    (h_phi_r, ROOT.BLUE, 20),
    (h_phi_t, ROOT.MAGENTA, 24),
):
    hist.SetLineColor(color)
    hist.SetFillColorAlpha(color, 0.7)
    hist.SetMarkerColor(color)
    hist.SetMarkerStyle(marker)
    hist.SetMarkerSize(0.7)
    hist.SetLineWidth(2)

for hist in (h_theta_r, h_theta_t, h_phi_r, h_phi_t):
    integral = hist.Integral("width")
    if integral > 0:
        hist.Scale(1.0 / integral)

c_theta_phi = ROOT.TCanvas("c_compare_theta_phi", "theta and phi distributions", 1200, 1600)
c_theta_phi.Divide(1,2)

c_theta_phi.cd(1)
h_theta_r.SetTitle("Reconstructed vs Generated Polar Angle")
h_theta_r.GetXaxis().SetTitle("#theta [rad]")
h_theta_r.GetYaxis().SetTitle("Normalized entries")
h_theta_r.Draw("hist")
h_theta_t.Draw("hist same")

theta_fd_min = 0.087
theta_fd_max = 0.611
y_theta_max = 1.05 * max(h_theta_r.GetMaximum(), h_theta_t.GetMaximum())
fd_line_color = ROOT.TColor.GetColor(150, 150, 150)
fd_lines_theta = []
for theta in (theta_fd_min, theta_fd_max):
    line = ROOT.TLine(theta, 0.0, theta, y_theta_max)
    line.SetLineColorAlpha(fd_line_color, 0.55)
    line.SetLineStyle(2)
    line.SetLineWidth(2)
    line.Draw()
    fd_lines_theta.append(line)

fd_label_theta = ROOT.TLatex()
fd_label_theta.SetTextAlign(22)
fd_label_theta.SetTextColorAlpha(fd_line_color, 0.75)
fd_label_theta.SetTextSize(0.035)
fd_label_theta.DrawLatex(
    0.5 * (theta_fd_min + theta_fd_max),
    0.98 * y_theta_max,
    "Forward Detector (FD)",
)

legend_theta = ROOT.TLegend(0.60, 0.70, 0.90, 0.90)
legend_theta.SetBorderSize(0)
legend_theta.AddEntry(h_theta_r, "CLAS12 Recon", "f")
legend_theta.AddEntry(h_theta_t, "RGA Clasdis Gen", "f")
legend_theta.Draw()

c_theta_phi.cd(2)
h_phi_r.SetTitle("Reconstructed vs Generated Azimuthal Angle")
h_phi_r.GetXaxis().SetTitle("#font[1]{#phi} [rad]")
# h_phi_r.GetYaxis().SetTitle("Normalized entries")
h_phi_r.GetYaxis().SetRangeUser(0,0.3)
h_phi_r.Draw("hist")
h_phi_t.Draw("hist same")

legend_phi_side = ROOT.TLegend(0.65, 0.75, 0.90, 0.90)
legend_phi_side.SetBorderSize(0)
legend_phi_side.AddEntry(h_phi_r, "CLAS12 Recon", "f")
legend_phi_side.AddEntry(h_phi_t, "RGA Clasdis Gen", "f")
# legend_phi_side.Draw()

c_theta_phi.Draw()
c_theta_phi.SaveAs("compare_theta_phi_notebook.png")
c_theta_phi.SaveAs("compare_theta_phi_notebook.pdf")
```
